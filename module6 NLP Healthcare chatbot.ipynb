```python
#import libraries
from transformers import pipeline

from deep_translator import GoogleTranslator

import re
```

```python
#Load NLP Pipelines
#Sentiment Pipeline
sentiment_pipe = pipeline(

    "sentiment-analysis",

    model="distilbert-base-uncased-finetuned-sst-2-english"
)
```

```python
#Lightweight Text Generation
chatbot_pipe = pipeline(

    "text-generation",

    model="distilgpt2"
)
```

```python
#Translation function 
def translate_text(

    text,

    target='en'
):

    translated = GoogleTranslator(

        source='auto',

        target=target

    ).translate(text)

    return translated
```

```python
#Sentiment Analysis
def analyze_sentiment(text):

    result = sentiment_pipe(text)

    return result
```

```python
#Healthcare Symptom Detection
symptom_keywords = [

    "fever",
    "cough",
    "chest pain",
    "fatigue",
    "headache",
    "dizziness",
    "shortness of breath",
    "palpitations",
    "vomiting",
    "diarrhea",
    "anxiety"
]
```

```python
#Extractor
def extract_symptoms(text):

    found = []

    text = text.lower()

    for symptom in symptom_keywords:

        if symptom in text:

            found.append(symptom)

    return found
```

```python
#Healthcare chatbot response
def healthcare_chatbot(text):

    symptoms = extract_symptoms(text)

    if "fever" in symptoms and "cough" in symptoms:

        return (
            "Respiratory symptoms detected. "
            "Please consult a healthcare professional "
            "if symptoms persist."
        )

    elif "chest pain" in symptoms:

        return (
            "Chest pain detected. "
            "Immediate medical evaluation is recommended."
        )

    else:

        return (
            "Symptoms noted. "
            "Please monitor your condition carefully."
        )
```

```python
#Test chatbot
sample_text = """

I have fever and cough
with fatigue for two days

"""

result = healthcare_chatbot(
    sample_text
)

print(result)
```

```python
#Multilingual Test
tamil_text = "எனக்கு காய்ச்சல் மற்றும் இருமல் உள்ளது"

translated = translate_text(
    tamil_text
)

print(translated)
```

```python
#Chatbot generation
response = chatbot_pipe(

    "Patient has chest pain and fatigue",

    max_length=50,

    num_return_sequences=1
)

print(response)
```
