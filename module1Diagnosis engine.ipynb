```python

selected_diseases = [
    'Asthma', 'Pneumonia', 'COPD', 'Flu',
    'Type2_Diabetes', 'Hyperthyroidism', 'Hypothyroidism', 'Obesity',
    'Irritable_Bowel_Syndrome', 'Gastritis', 'Peptic_Ulcer_Disease',
    'Chronic_Kidney_Disease', 'Hypertension', 'Osteoarthritis'
]

# Create a filtered copy without changing the original dataset
df_model = df[df['disease'].isin(selected_diseases)].copy()

print("Filtered Dataset Shape:", df_model.shape)
print("\nSelected Disease Distribution:")
print(df_model['disease'].value_counts())
```

```python

#Separate input features and target

X = df_model.drop(columns=["disease"])
y = df_model["disease"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nNumber of Disease Classes:", y.nunique())

print("\nNon-numeric Columns:")
print(X.select_dtypes(exclude=np.number).columns.tolist())

print("\nMissing Values:", X.isnull().sum().sum())

print("\nDuplicate Rows:", df_model.duplicated().sum())
```

```python

from sklearn.model_selection import train_test_split

# Split dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("\nTraining Target:", y_train.shape)
print("Testing Target:", y_test.shape)

print("\nTraining Disease Distribution:")
print(y_train.value_counts().sort_index())

print("\nTesting Disease Distribution:")
print(y_test.value_counts().sort_index())
```

```python
#import libraries

from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=20,
    min_samples_split=5,
    min_samples_leaf=2,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

print("Random Forest model initialized successfully.")
```

```python
#Train Random Forest

rf_model.fit(X_train, y_train)

print("Random Forest training completed successfully.")
```

```python
#Cross-validation


from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = cross_validate(
    estimator=rf_model,
    X=X_train,
    y=y_train,
    cv=cv,
    scoring=["accuracy", "f1_macro", "f1_weighted"],
    n_jobs=1
)

print("Fold Accuracy:", cv_results["test_accuracy"])
print("Mean Accuracy:", cv_results["test_accuracy"].mean())

print("\nFold Macro F1:", cv_results["test_f1_macro"])
print("Mean Macro F1:", cv_results["test_f1_macro"].mean())

print("\nFold Weighted F1:", cv_results["test_f1_weighted"])
print("Mean Weighted F1:", cv_results["test_f1_weighted"].mean())
```

```python
#Encode the target labels

from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

print("Number of classes:", len(label_encoder.classes_))
print("Training shape:", y_train_encoded.shape)
print("Testing shape:", y_test_encoded.shape)
```

```python
#RANDOM FOREST EVALUATION

from sklearn.metrics import accuracy_score, f1_score

#Generate predictions on test data
y_rf_pred = rf_model.predict(X_test)

print("RANDOM FOREST HOLDOUT EVALUATION")
print("=" * 45)

print("Training Accuracy:",
      round(rf_model.score(X_train, y_train), 4))

print("Testing Accuracy:",
      round(accuracy_score(y_test, y_rf_pred), 4))

print("Macro F1:",
      round(f1_score(y_test, y_rf_pred, average="macro"), 4))

print("Weighted F1:",
      round(f1_score(y_test, y_rf_pred, average="weighted"), 4))
```

```python
#Initialize XGBoost

from xgboost import XGBClassifier

xgb_model = XGBClassifier(
    objective="multi:softprob",
    num_class=14,
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=2,
    eval_metric="mlogloss",
    random_state=42,
    n_jobs=-1
)

print("XGBoost model initialized successfully.")
```

```python
#XGBoost cross-validation

# ============================================
# XGBOOST - 5-FOLD CROSS-VALIDATION
# ============================================

from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

xgb_cv_results = cross_validate(
    estimator=xgb_model,
    X=X_train,
    y=y_train_encoded,
    cv=cv,
    scoring={
        "accuracy": "accuracy",
        "f1_macro": "f1_macro",
        "f1_weighted": "f1_weighted"
    },
    n_jobs=1
)

# Display results
print("XGBOOST CROSS-VALIDATION RESULTS")
print("=" * 45)

for i in range(5):
    print(f"\nFold {i + 1}")
    print("Accuracy:", round(xgb_cv_results["test_accuracy"][i], 4))
    print("Macro F1:", round(xgb_cv_results["test_f1_macro"][i], 4))
    print("Weighted F1:", round(xgb_cv_results["test_f1_weighted"][i], 4))

# Average scores
print("\nAVERAGE SCORES")
print("=" * 45)

print("Mean Accuracy:",
      round(xgb_cv_results["test_accuracy"].mean(), 4))

print("Mean Macro F1:",
      round(xgb_cv_results["test_f1_macro"].mean(), 4))

print("Mean Weighted F1:",
      round(xgb_cv_results["test_f1_weighted"].mean(), 4))
```

```python

# ============================================
# FINAL XGBOOST MODEL EVALUATION
# ============================================

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

#Step 1:Train the model on the complete training set
              
xgb_model.fit(X_train, y_train_encoded)

# Step 2: Predict training and testing data
y_train_pred = xgb_model.predict(X_train)
y_test_pred = xgb_model.predict(X_test)

# Step 3: Training performance
print("TRAINING PERFORMANCE")
print("=" * 45)

print("Training Accuracy:",
      round(accuracy_score(y_train_encoded, y_train_pred), 4))

# Step 4: Testing performance
print("\nTESTING PERFORMANCE")
print("=" * 45)

print("Testing Accuracy:",
      round(accuracy_score(y_test_encoded, y_test_pred), 4))

print("Macro F1:",
      round(f1_score(y_test_encoded, y_test_pred, average="macro"), 4))

print("Weighted F1:",
      round(f1_score(y_test_encoded, y_test_pred, average="weighted"), 4))

# Step 5: Classification report
print("\nCLASSIFICATION REPORT")
print("=" * 65)

print(classification_report(
    y_test_encoded,
    y_test_pred,
    labels=range(len(label_encoder.classes_)),
    target_names=label_encoder.classes_,
    zero_division=0
))

# Step 6: Confusion matrix
cm = confusion_matrix(
    y_test_encoded,
    y_test_pred,
    labels=range(len(label_encoder.classes_))
)

print("Confusion Matrix Shape:", cm.shape)
print(cm)
```

```python
# Correlation between numerical features
correlation_matrix = X_train.corr().abs()

import numpy as np

upper_triangle = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
)

high_correlations = (
    upper_triangle.stack()
    .reset_index()
)

high_correlations.columns = [
    "Feature 1", "Feature 2", "Correlation"
]

high_correlations = high_correlations[
    high_correlations["Correlation"] >= 0.90
].sort_values(by="Correlation", ascending=False)

print("Highly correlated feature pairs:")
print(high_correlations.head(30).to_string(index=False))
```

```python
#feature selection
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectFromModel
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_validate, StratifiedKFold

selector_pipeline = Pipeline([
    ("feature_selection", SelectFromModel(
        RandomForestClassifier(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ),
        threshold="median"
    )),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        max_depth=20,
        min_samples_split=5,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

selector_scores = cross_validate(
    selector_pipeline,
    X_train,
    y_train,
    cv=cv,
    scoring={
        "accuracy": "accuracy",
        "macro_f1": "f1_macro",
        "weighted_f1": "f1_weighted"
    },
    n_jobs=1
)

print("Feature Selection - 5 Fold Results")
print("Mean Accuracy:", selector_scores["test_accuracy"].mean())
print("Mean Macro F1:", selector_scores["test_macro_f1"].mean())
print("Mean Weighted F1:", selector_scores["test_weighted_f1"].mean())
```

```python
#Tune Random Forest
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_validate, StratifiedKFold

tuned_rf = RandomForestClassifier(
    n_estimators=300,
    max_depth=12,
    min_samples_split=10,
    min_samples_leaf=5,
    max_features="sqrt",
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

tuned_scores = cross_validate(
    tuned_rf,
    X_train,
    y_train,
    cv=cv,
    scoring={
        "accuracy": "accuracy",
        "macro_f1": "f1_macro",
        "weighted_f1": "f1_weighted"
    },
    n_jobs=1
)

print("Tuned Random Forest - 5 Fold Results")
print("Mean Accuracy:", tuned_scores["test_accuracy"].mean())
print("Mean Macro F1:", tuned_scores["test_macro_f1"].mean())
print("Mean Weighted F1:", tuned_scores["test_weighted_f1"].mean())
```
